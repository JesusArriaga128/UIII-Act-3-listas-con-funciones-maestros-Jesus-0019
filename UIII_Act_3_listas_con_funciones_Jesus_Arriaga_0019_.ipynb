### Lista (Maestros)


```python
# Definimos la clase para los maestros
class Maestro:
    def __init__(self, id_m, nombre, materia):
        self.id_m = id_m
        self.nombre = nombre
        self.materia = materia

    def __str__(self):
        return f"ID: {self.id_m} | Maestro: {self.nombre} | Materia: {self.materia}"

# Lista para almacenar los datos (nuestra base de datos)
lista_maestros = []

# --- Definición de las funciones solicitadas ---

def agregarmaestro():
    print("\n--- Agregar Nuevo Maestro ---")
    id_m = input("ID: ")
    nombre = input("Nombre: ")
    materia = input("Materia: ")
    nuevo = Maestro(id_m, nombre, materia)
    lista_maestros.append(nuevo)
    print("¡Maestro agregado con éxito!")

def leermaestro():
    print("\n--- Consultar Maestro ---")
    id_buscado = input("ID a buscar: ")
    for m in lista_maestros:
        if m.id_m == id_buscado:
            print(m)
            return
    print("Maestro no encontrado.")

def listarmaestro():
    print("\n--- Lista de Maestros ---")
    if not lista_maestros:
        print("No hay registros en la lista.")
    else:
        for m in lista_maestros:
            print(m)

def buscarmaestro():
    print("\n--- Buscar Maestro por Nombre ---")
    nom = input("Nombre a buscar: ").lower()
    encontrados = [m for m in lista_maestros if nom in m.nombre.lower()]
    if encontrados:
        for m in encontrados:
            print(m)
    else:
        print("No se encontraron coincidencias.")

def modificamaestro():
    print("\n--- Modificar Datos de Maestro ---")
    id_mod = input("ID del maestro a editar: ")
    for m in lista_maestros:
        if m.id_m == id_mod:
            m.nombre = input(f"Nuevo nombre [{m.nombre}]: ") or m.nombre
            m.materia = input(f"Nueva materia [{m.materia}]: ") or m.materia
            print("Datos actualizados.")
            return
    print("ID no encontrado.")

def borrarmaestro():
    print("\n--- Borrar Maestro ---")
    id_del = input("ID a eliminar: ")
    for i, m in enumerate(lista_maestros):
        if m.id_m == id_del:
            lista_maestros.pop(i)
            print("Registro eliminado.")
            return
    print("ID no encontrado.")

# --- Menú Principal con While True ---

def menu_principal():
    while True:
        print("\n===============================")
        print("    SISTEMA DE PROFESORES")
        print("===============================")
        print("1. Agregar profesor")
        print("2. Leer/Ver profesor")
        print("3. Listar todos")
        print("4. Buscar profesor")
        print("5. Modificar profesor")
        print("6. Salir")
        print("===============================")

        opcion = input("Seleccione una opción (1-6): ")

        if opcion == "1":
            agregarmaestro()
        elif opcion == "2":
            leermaestro()
        elif opcion == "3":
            listarmaestro()
        elif opcion == "4":
            buscarmaestro()
        elif opcion == "5":
            modificamaestro()
        elif opcion == "6":
            print("\nSaliendo del sistema... ¡Hasta pronto!")
            print("------------------------------------------")
            print("Programa realizado por Jesus Arriaga NL = 5")
            print("------------------------------------------")
            break
        else:
            print("\n[!] Opción no válida. Por favor, intente de nuevo.")

if __name__ == "__main__":
    menu_principal()
```
