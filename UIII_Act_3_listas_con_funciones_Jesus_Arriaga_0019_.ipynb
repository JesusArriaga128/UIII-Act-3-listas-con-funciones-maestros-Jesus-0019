### Lista (Maestros)


```python
def menu_principal():
    # Simulamos una base de datos simple con una lista
    profesores = []

    while True:
        print("\n===============================")
        print("    SISTEMA DE PROFESORES")
        print("===============================")
        print("1. Agregar profesor")
        print("2. Leer/Ver profesor")
        print("3. Listar todos")
        print("4. Buscar profesor")
        print("5. Modificar profesor")
        print("6. Salir")
        print("===============================")

        opcion = input("Seleccione una opción (1-6): ")

        if opcion == "1":
            print("\n-> Ejecutando agregarmaestro()...")
            id_p = input("Ingrese ID: ")
            nombre = input("Ingrese Nombre: ")
            materia = input("Ingrese Materia: ")
            # Guardamos como un diccionario para facilitar el manejo
            profesores.append({"id": id_p, "nombre": nombre, "materia": materia})
            print("Profesor agregado.")

        elif opcion == "2":
            print("\n-> Ejecutando leermaestro()...")
            id_buscado = input("Ingrese ID a consultar: ")
            encontrado = False
            for p in profesores:
                if p["id"] == id_buscado:
                    print(f"Datos: {p}")
                    encontrado = True
                    break
            if not encontrado: print("No se encontró el registro.")

        elif opcion == "3":
            print("\n-> Ejecutando listarmaestro()...")
            if not profesores:
                print("La lista está vacía.")
            for p in profesores:
                print(f"ID: {p['id']} | Nombre: {p['nombre']} | Materia: {p['materia']}")

        elif opcion == "4":
            print("\n-> Ejecutando buscarmaestro()...")
            nom_buscar = input("Ingrese nombre a buscar: ").lower()
            encontrados = [p for p in profesores if nom_buscar in p["nombre"].lower()]
            if encontrados:
                for p in encontrados: print(p)
            else:
                print("Sin coincidencias.")

        elif opcion == "5":
            print("\n-> Ejecutando modificamaestro()...")
            id_mod = input("Ingrese ID del profesor a modificar: ")
            for p in profesores:
                if p["id"] == id_mod:
                    p["nombre"] = input(f"Nuevo nombre ({p['nombre']}): ") or p["nombre"]
                    p["materia"] = input(f"Nueva materia ({p['materia']}): ") or p["materia"]
                    print("Registro actualizado.")
                    break
            else:
                print("ID no encontrado.")

        elif opcion == "6":
            print("\nSaliendo del sistema... ¡Hasta pronto!")
            print("******************************************")
            print("Programa realizado por Jesus Arriaga NL = 5")
            print("******************************************")
            break  # Este comando rompe el bucle while True

        else:
            print("\n[!] Opción no válida. Por favor, intente de nuevo.")

if __name__ == "__main__":
    menu_principal()
```
